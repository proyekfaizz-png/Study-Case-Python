**PROJECT**
**ANALISIS DATA PENJUALAN TOKO ONLINE**

1. Langkah 1 (Total Penjualan):
Buat kolom baru bernama 'Total_Pendapatan' yang berisi perkalian antara 'Harga_Satuan' dan 'Jumlah_Terjual'.

2. Langkah 2 (Filtering Produk Laris):
Tampilkan daftar transaksi yang 'Jumlah_Terjual'-nya lebih dari atau sama dengan 5.

3. Langkah 3 (GroupBy Per Kategori):
Hitung total gabungan 'Total_Pendapatan' untuk masing-masing 'Kategori' (Gunakan .sum(), bukan .mean()).

4. Langkah 4 (Sorting):
Urutkan hasil dari Langkah 3 berdasarkan pendapatannya dari yang tertinggi ke terendah.

```python
import pandas as pd

data_toko = {
    'Nama_Produk': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Mouse', 'Laptop', 'Keyboard'],
    'Kategori': ['Elektronik', 'Aksesori', 'Aksesori', 'Elektronik', 'Aksesori', 'Elektronik', 'Aksesori'],
    'Harga_Satuan': [8000000, 150000, 300000, 2000000, 150000, 8000000, 300000],
    'Jumlah_Terjual': [2, 10, 5, 3, 8, 1, 4]
}

df_toko = pd.DataFrame(data_toko)

#1. membuat kolom baru "total pendapatan"
df_toko['Total_Pendapatan'] = df_toko['Harga_Satuan'] * df_toko['Jumlah_Terjual']

#2. filter produk jumlah terjual >=5
filter_produk = df_toko[df_toko['Jumlah_Terjual'] >= 5]
print("=== FILTER PRODUK DENGAN JUMLAH TERJUAL (>=5)")
print(filter_produk[['Nama_Produk', 'Jumlah_Terjual', 'Total_Pendapatan']])

#3. group per kategori
total_per_kategori = df_toko.groupby('Kategori')['Total_Pendapatan'].sum()

#4. mengurutkan dari yang teratas
df_sorted = total_per_kategori.sort_values(ascending=False)
print("\n================================")
print("TOTAL PENDAPATAN PER KATEGORI")
print(df_sorted)
```
